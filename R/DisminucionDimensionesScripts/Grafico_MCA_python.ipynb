```python
import matplotlib.pyplot as plt

# Obtener las coordenadas de las columnas
column_coords = mca_2.column_coordinates(data_object[['Type', 'Content Rating']])

cont = 0
j = 0

# Lista de colores para asignar a cada punto del nombre
colores = ['red', 'green', 'blue', 'orange', 'purple', 'yellow', 'pink', 'brown']  # Añadir más colores si es necesario

while cont < 4:
    # Extraer los nombres y las coordenadas
    nombres = column_coords.index
    coordenadas_x = column_coords[j]
    coordenadas_y = column_coords[j+1]
    
    # Crear el gráfico de dispersión
    plt.figure(figsize=(8, 6))  # Tamaño de la figura
    
    # Diccionario para mapear cada nombre a su color correspondiente
    color_map = {nombre: colores[i % len(colores)] for i, nombre in enumerate(nombres)}
    
    # Agregar etiquetas a cada punto con colores diferentes y registrar el color en la leyenda
    for nombre, x, y in zip(nombres, coordenadas_x, coordenadas_y):
        color = color_map[nombre]  # Obtener el color del diccionario
        plt.scatter(x, y, color=color, label=nombre)  # Pintar el punto con un color diferente y registrar el nombre en la leyenda
        plt.annotate(nombre, (x, y), textcoords="offset points", xytext=(0,10), ha='center', color='black')  # Mantener el color del texto del nombre en negro
    
    # Etiquetas de los ejes
    plt.xlabel(f'Componente {j+1}')
    plt.ylabel(f'Componente {j+2}')
    
    # Título del gráfico
    plt.title('Gráfico de Dispersión CA')
    
    # Mostrar la leyenda
    plt.legend()
    
    # Mostrar el gráfico
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    cont += 1
    j += 1
```
